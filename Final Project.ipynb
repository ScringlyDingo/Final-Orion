```python
#Keno gambling game
#references for myself..
#& - finds numbers shared between two sets
#try/except ValueError - deals with numbers that can't be converted into integers
#/n - new line
import random 

print("Casino gambling experience: KENO")
print("Pick five numbers from 1-80...")



#retrieve the player numbers 
while True: 
    try:
        choices = input("Enter five different numbers seperated with spaces: ")
        player_number_choices = [int(x) for x in choices.split()]

        if len(player_number_choices) != 5:
            print("Enter five numbers.")
            continue 

        if len(set(player_number_choices)) != 5:
            print("‼️Those numbers will not work‼️.")
            continue 

        if any(n < 1 or n > 80 for n in player_number_choices):
            print("The numbers have to be in between 1 and 80.")
            continue  

        break 

    except ValueError:
        print("Amigo only enter numbers.") 





#20 random numbers
drawn_numbers = random.sample(range(1, 81), 20) 

#,atching with the player numbers
matches = sorted(set(player_number_choices) & set(drawn_numbers))

#the results 
print("\nYour number choices:", sorted(player_number_choices))
print("Drawn numbers:", sorted(drawn_numbers))
print("your matches:", matches)
print("Number of matches:", len(matches))



payouts = {
    0: 0,
    1: 0,
    2: 0.00000000001,
    3: 5,
    4: 25,
    5: 100
} 
print("The money you recieve:", payouts[len(matches)], "points")
if len(matches) == 5:
    print("You can afford a private jet!") 
elif len(matches) >= 3: 
    print("I've seen worse") 
else:
    print("😂😂😂") 
```
